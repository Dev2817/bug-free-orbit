# Changing string variables and concatenating strings

```python
# Strings are immutable and cannot be changed.
  #Individual characters of a string cannot be changed.

alphabet = 'abcdefghijklmnopqrstuvwxyz'

# change to uppercase
alphabet[0] = 'A'  # This will raise an error
alphabet[1] = 'B'  # This will raise an error
alphabet[25] = 'Z'  # This will raise an error

print('Alphabet:', alphabet)  # This will print the original string unchanged
```

```python
# Instead, update the variable by assigning an entirely new string.

alphabet = 'abcdefghijklmnopqrstuvwxyz'

# change to uppercase

alphabet = 'ABCDEFGHIJKLMNOPQRSTUVWXYZ'  # Assign a new string to the variable

print('Alphabet:', alphabet)  # This will print the updated string
```

```python
# String concatenation

first_name = "John"
last_name = "Doe"

full_name = first_name + " " + last_name

print("Full Name:", full_name)  # Output: JohnDoe
```

# List

```python
# Access list elements using an indexing expression.

Lamborgini_urus = 3900000 #  3.9 million
Bugatti_Chiron = 3000000 #  3 million
BMW_M5 = 7000000 #  7 million 

prices = [Lamborgini_urus, Bugatti_Chiron, BMW_M5]

print('Lamborgini_urus:', prices[0], '$')
print('Bugatti_Chiron:', prices[1], '$')
print('BMW_M5:', prices[2], '$')
```

```python
# Updating list elements

my_favorite_games = ['The Witcher 3', 'Cyberpunk 2077', 'Elden Ring']
print(my_favorite_games)

# update a list element
my_favorite_games[1] = 'God of War Ragnarok'
print(my_favorite_games)
```

```python
# Adding and removing list elements

my_games = ['Assassins Creed Origins', 'Minecraft']
print(my_games)

# append a new game to the list
my_games.append('God of War')
print(my_games)

# pop a game from the list
removed_game = my_games.pop(2)
print('Removed Game:', removed_game)
print('Updated Games List:', my_games)

# remove a game by name
my_games.remove('Assassins Creed Origins')
print('Games List after removal:', my_games)
```

```python
# Sequence-type methods and functions

# concatenating  lists
house_prices = [250000, 300000, 400000] + [150000]
print('There are', len(house_prices), 'houses available.')

# finding min and max prices
print('The cheapest house is priced at $', min(house_prices))
print('The most expensive house is priced at $', max(house_prices))
```

# Set Basics

```python
# Create a set using set() function

A = set([1,7,4])

# creating a set using set literal

B = {2,8,6}

# PRINT THE CONTENT OF THE SET

print('Set A:', A)
print('Set B:', B)
```

## Modifying Sets

```python
# create sets

names1 = {'Loki', 'Thor', 'Odin'}
names2 = {'Freya', 'Kratos', 'Balder'}

# Add elements to a set using add() method

names1.add('Frigg')
names2.add('Tyr')
print('Names Set 1:', names1)
print('Names Set 2:', names2)
```

```python
# updating sets using update() method

names1.update(names2)
print('Updated Names Set 1:', names1)
```

```python
# removing elements from a set using remove() method

names1.remove('Frigg')
names2.remove('Freya')
print('Names Set 1 after removal:', names1)
print('Names Set 2 after removal:', names2)
```

```python
# clearing a set using clear() method

names1.clear()
print('Names Set 1 after clearing:', names1)
```

# Set theory operations.

```python
game1 = {'kratos', 'freya', 'balder','loki'}
game2 = {'loki', 'thor', 'odin'}
game3 = {'osirus', 'anubis', 'sobek','loki'}
game4 = {'steve, alex, herobrine','loki'}

# union game 1 and game 2
result_game_set = game1.union(game2)

print('Union of Game 1 and Game 2:', result_game_set)
```

```python
# interesection between result_game_set and game3

common_characters = result_game_set.intersection(game3)
print('Common characters between Result Game Set and Game 3:', common_characters)
```

```python
# difference between result_game_set and game4

difference_characters = result_game_set.difference(game4)
print('Characters in Result Game Set but not in Game 4:', difference_characters)
```

```python
# symmetric difference between result_game_set and game4

symmetric_difference_characters = result_game_set.symmetric_difference(game4)
print('Characters in either Result Game Set or Game 4 but not in both:', symmetric_difference_characters)
```

#  Dictionary basics

```python
# Creating a dictionary.

players = {'Dev-OG': 'Gamer and streamer', 
           'Hunter-OG': 'Editor and content creator'
}
print ('Players Dictionary:', players)
```

```python
# Accessing dictionary entries

prices = {'Notch apple': 1000, 'Diamond sword': 950, 'Enchanted bow': 900}
print('Price of Notch apple:', prices['Notch apple'], '$')

print('Price of Diamond Armour:', prices['Diamond Armour'], '$') # is not present in the dictionary

```

```python
# Adding, modifying, and removing dictionary entries

# adding a new entry
players = {}
players ['Dev-OG'] = 'Gamer and streamer' # New entry
print('Players Dictionary after adding Dev-OG:', players)
```

```python
# modifying an existing entry

players ['Dev-OG'] = 'Owner of YouTube channel' # Modified entry
print('Players Dictionary after modifying Dev-OG:', players)
```

```python
# deleting an entry

del players ['Dev-OG']  # Deleting the entry
print(players)
```
