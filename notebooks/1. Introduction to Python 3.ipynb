# Logic Error

```python
# fixed version

current_salary = int(input('Enter current salary: '))

# give the salary hike by 5%

raised_salary = 0.05
new_salary = current_salary + (current_salary * raised_salary)
print('current salary is:', current_salary)
print('New salary is:', new_salary)
```

```python
# error version

current_salary = int(input('Enter current salary: '))

# give the salary hike by 5%

raised_salary = 5 #logic error gives 500% hike instead of 5% so the correct value is 0.05
new_salary = current_salary + (current_salary * raised_salary)
print('current salary is:', current_salary)
print('New salary is:', new_salary)
```
