# Variables and Expressions

 Printing displays an object's value

```python
# Printing displays an object's value

x = 2+2
print(x)
print(5)
```

 Using type() to print an object's type

```python
# Using type() to print an object's type

x = 2+2
print(type(x))
print(type('ABC'))
```

Using id() to print an object's identity.

```python
# Using id() to print an object's identity.

x = 2 + 2
print(id(x))
print(id('ABC'))
```

```python
print(2**2) # Exponentiation

print(2e2) # Scientific notation (2 x 10^2)
```

```python
""" Compute the total cost of leasing a car given the down payment,
monthly rate and number of months. """

down_payment = int(input("Enter the down payment: ")) # Input for down payment(from user)
print("Down payment is:", down_payment)

payment_per_month = int(input("Enter the monthly rate: ")) # Input for monthly rate (from user)
print("Monthly rate is:", payment_per_month)

num_months = int(input("Enter the number of months: ")) # Input for number of months (from user)
print("Number of months is:", num_months)

total_cost = down_payment + (payment_per_month * num_months)
print("The total cost of leasing the car is:", total_cost)
```

#  Division and modulo

```python
print(23/10)  # Division
print(20//10) # Floor Division
print(7%3)    # Modulus (REMAINDER)
print(51%2)    # Modulus when dividend is smaller than divisor (REMAINDER)
print(100%(2//2)) # Division by zero (results in error)
```

```python
# lets take a 3 digit number

user_value = int(input("Enter a 3 digit number: "))
print("You entered:", user_value)

onces_digit = user_value % 10
temp_value = user_value // 10
print("Onces digit:", onces_digit)

tens_digit = temp_value % 10
temp_value = temp_value // 10
print("Tens digit:", tens_digit)

hundreds_digit = temp_value % 10
print("Hundreds digit:", hundreds_digit)
```

```python
""" Given a 10-digit phone number stored as an integer,
% and // can be used to get any part, such as the prefix. 
 or phone_num = 9365551212 (whose prefix is 555):"""

phone_num = 9365551212
tmp_val = phone_num // 10000   # // 10000 shifts right by 4, so 936555
prefix_num = tmp_val % 1000    # % 1000 gets the right 3 digits, so 555.

print("The phone number is:", phone_num)
print("The tmp value is:", tmp_val)
print("The prefix number is:", prefix_num)
```
