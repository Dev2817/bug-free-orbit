# Branching (if - elif - else)

```python
# if statement

hotel_rate = 150 # Base rate per night in dollars
num_years = int(input("Enter Married years: "))
print('Entered years of Marriage:', num_years)

if num_years == 50:
    print("Congratulations on your Golden Anniversary!.")
    hotel_rate = hotel_rate / 2

print (f'your hotel rate is: ${hotel_rate:.2f}')
```

```python
# If-else statement

user_num = int(input("Enter a number: "))

div_remainder = user_num % 2

if div_remainder == 0:
    print(f"{user_num} is an even number.")
else:
    print(f"{user_num} is an odd number.")
```

```python
# if-elif-else statement

num_years = int(input("Enter Married years: "))

if num_years == 1:
    print("Happy first anniversary!")
elif num_years == 10:
    print("Happy tenth anniversary!")
elif num_years == 25:
    print("Happy silver anniversary!")
elif num_years == 50:
    print("Happy golden anniversary!")
else:
    print("Congratulations on your anniversary!")
```

```python
# Detecting ranges with branches

user_age = int(input("Enter your age: "))
print('Entered age:', user_age)

if user_age < 16:   # age 15 and below
    print("to young")
    insurance_price = 0
elif user_age < 25: # age 16 to 24
    insurance_price = 5000
elif user_age < 40: # age 25 to 39
    insurance_price = 3000
else:               # age 40 and above
    insurance_price = 2000

print(f"Your insurance price is: ${insurance_price}")
```

```python
# Detecting ranges: Cable TV channels.

cable_channel = int(input("Enter cable channel number: "))
print('Entered channel number:', cable_channel)

if (cable_channel >= 2) and (cable_channel <= 199):
    print("You are watching local channels.")
elif (cable_channel >= 502) and (cable_channel <= 599):
    print("You are watching sports channels.")
elif (cable_channel >= 901) and (cable_channel <= 999):
    print("You are watching movie channels.")
else:
    print("Invalid channel number.")
```

```python
# Detecting ranges implicitly vs. explicitly.
## explicit range detection

x = int(input("Enter a number: "))
print('Entered number:', x)

if x < 0:
    print("x is negative")
elif (x >= 0) and (x <= 10):
    print("x is between 0 and 10")
elif (x>= 11) and (x <= 20):
    print("x is between 11 and 20")
else:
    print("x is greater than 20")
```

```python
# implicit range detection

x = int(input("Enter a number: "))
print('Entered number:', x) 

if x < 0:
    print("x is negative")  
elif x <= 10:
    print("x is between 0 and 10")
elif x <= 20:
    print("x is between 11 and 20")
else:
    print("x is greater than 20")
```

```python
# Basic ranges with gaps

movie_ticket_price = None
age = int(input("Enter your age: "))
print('Entered age:', age)

if age <= 12:
    print("Child ticket price applied.")
    movie_ticket_price = 80
elif age >= 65:
    print("Senior ticket price applied.")
    movie_ticket_price = 100
else:
    print("Adult ticket price applied.")
    movie_ticket_price = 150

print(f"Your movie ticket price is: ₹{movie_ticket_price}")
```

```python
# Explicit ranges with gaps detection using logical AND.

office_id = int(input("Enter your office ID: "))
print('Entered office ID:', office_id)

if office_id >= 100 and office_id <= 150: 
    print("valid office ID")
elif office_id >= 200 and office_id <= 250:
    print("valid office ID")
else:
    print("invalid office ID")  
```

```python
# # Explicit ranges with gaps detection using logical OR.

office_id = int(input("Enter your office ID: "))
print('Entered office ID:', office_id)  

if (office_id >= 100 and office_id <= 150) or (office_id >= 200 and office_id <= 250):
    print("valid office ID")
else:
    print("invalid office ID")
```

```python
# Multiple distinct if statements

user_age = int(input("Enter your age: "))
print('Entered age:', user_age)

# Note that more than one if statement can execute
if user_age <= 13:
    print("You are too young.")   
if user_age > 17: 
    print("You are old enough to drive.")  
if user_age > 18:
    print("You are old enough to vote.")
if user_age >= 21:
    print("You are old enough to drink alcohol.")   
if user_age >= 65:
    print("You are a senior citizen.")  
```

```python
# Nested if-else statements 

x = 30
y = 20

if x >= y:
    print("x is greater than or equal to y")
    if x == y:
        print("x is equal to y")
    else:
        print("x is greater than y")
else:
    print("x is less than y")
```

```python
# Membership operators: in/not in

## Using 'in' operator

my_list = [10, 20, 30, 40, 50]
value = int(input("Enter a value to search in the list: "))
print('Entered value:', value)
if value in my_list:
    print(f"{value} is found in the list.")
else:
    print(f"{value} is not found in the list.")
```

```python
# using 'not in' operator

my_list = [10, 20, 30, 40, 50]
value = int(input("Enter a value to search in the list: "))
print('Entered value:', value)
if value not in my_list:
    print(f"{value} is not found in the list.")
else:
    print(f"{value} is found in the list.")
```

```python
# Checking for membership in a dict.

my_dict = {'a': 1, 'b': 2, 'c': 3}

key = input("Enter a key to search in the dictionary: ")
print('Entered key:', key)
if key in my_dict:
    print(f"Key '{key}' is found in the dictionary with value: {my_dict[key]}")
else:
    print(f"Key '{key}' is not found in the dictionary.")
    
```
