# Functions

```python
# Function example: Printing a pizza area.

def calc_pizza_area():
   pizza_val = 3.14

   pizza_diamater = 12.0
   pizza_radius = pizza_diamater / 2.0
   pizza_area = pizza_val * pizza_radius * pizza_radius
   return pizza_area


print ("12 inch pizza area is: ", calc_pizza_area(), "square inches")
```

```python
# Function example: Returning a value.

def compute_square (number_to_square):
   return number_to_square * number_to_square

number_square = compute_square(5)
print ("The square of 5 is: ", number_square)
```

```python
# Function with multiple parameters.

def calc_pizza_volume (pizza_diameter, pizza_height):
    pi_value = 3.14

    pizza_radius = pizza_diameter / 2.0
    pizza_area = pi_value * pizza_radius * pizza_radius                                 
    pizza_volume = pizza_area * pizza_height             
    return pizza_volume
                                                            
```

```python
print('12.0 x 0.8 inch pizza is:',calc_pizza_volume (12.0,0.8),'cubic inches.')
```

```python
print('12.0 x 0.3 inch pizza is:',calc_pizza_volume (12.0,0.3),'cubic inches.')
```

```python
print('16.0 x 0.8 inch pizza is:',calc_pizza_volume (16.0,0.8),'cubic inches.')
```
