# While loop: Basics


```python
# While loop: Basics

power = 2
user_char = input("Enter either 'y' or 'n': ")

while user_char == 'y':
    print (power)
    power = power * 2
    user_char = input("Enter either 'y' or 'n': ")

print ("Goodbye!")
```

```python
# While loop example: Face-printing program that ends when user enters 'q'.

nose = '0' # looks little like a nose
user_val = '-'

while user_val != 'q':
    print (f' {user_val} {user_val} ')
    print (f'   {nose}   ')  # print nose
    print (user_val * 5)  # print mouth
    print ('\n')

    user_input = input("Enter a character to use for the face, or 'q' to quit: ")
    user_val = user_input[0] 
print ("Goodbye! \n")
```

```python
# Stepping through a while loop

num = int(input("Enter a number: "))
print ('entered number:', num)

# print each digit 
while num > 0:
    num = num % 10
    print (num)
    num = num // 10
    print (num)
```

```python
# 1) While loop example: Factorial program.

n = int(input("Enter a positive integer: "))
result = 1
i = 1

while i <= n:
    result = result * i
    i = i + 1

print (f'The factorial of {n} is {result}')
```

```python
# 1) Computing average of a list with a sentinel

total = 0
count = 0

user_input = int(input())

while user_input > 0:
    total = total + user_input
    count = count + 1
    user_input = int(input())

print (f'Average is: {total / count:.0f}\n')
```

```python
# Counting with a while loop

n = int(input("Enter a positive integer: "))
i = 1

print ("Entered Number:", n)

while i <= n:
    print (i)
    i = i + 1   
print ("Done!\n")
```

```python
# For loops
## A for loop assigns the loop variable with a dictionary's keys.

channels = {'movie': 1, 'news': 2, 'sports': 3, 'music': 4}

for c in channels:
    print (f'{c} is on Channel: {channels[c]}')
```

```python
# Using a for loop to access each character of a string.

str = ''
for character in "Hello":
    str = str + character + '-'
print (str) 
```

```python
# For loop example: Calculating shop revenue. using list of prices.

prices = [5.99, 15.49, 3.99, 29.99, 12.49]

total = 0
for price in prices:
    total = total + price
print (f'Total revenue is: ${total:.2f}')

# calculating average price 
average = total / len(prices)
print (f'Average price is: ${average:.2f}')
```

```python
# For loop example: Looping over a sequence in reverse.

names = ['Loki', 'Thor', 'Odin', 'heimdall']

for name in names:
    print (name, '|', end=' ')
print()  # for new line

print("\n Printing in Reverse Order:" )
for name in reversed(names):
    print (name, '|', end=' ')
```

```python
# nested loops
''' program to print all 2 letter damain names'''

print("2 letter domain names:")

letter1 = 'a'
letter2 = '?'

while letter1 <= 'z':
    letter2 = 'a'
    while letter2 <= 'z':
        print (f' {letter1 + letter2}.com')
        letter2 = chr(ord(letter2) + 1)
    letter1 = chr(ord(letter1) + 1)
```

```python
# Incremental programming

user_input = input('Enter phone number: ')
print (f'Entered phone number: {user_input}')
phone_number = ''

for char in user_input:
    if ('0'<= char <= '9') or (char == '-'):
        phone_number += char
    elif ('a' <= char <= 'c') or ('A' <= char <= 'C'): 
        phone_number += '2'
    elif ('d' <= char <= 'f') or ('D' <= char <= 'F'): 
        phone_number += '3'
    elif ('g' <= char <= 'i') or ('G' <= char <= 'I'):          
        phone_number += '4'
    elif ('j' <= char <= 'l') or ('J' <= char <= 'L'):          
        phone_number += '5'
    elif ('m' <= char <= 'o') or ('M' <= char <= 'O'):          
        phone_number += '6'
    elif ('p' <= char <= 's') or ('P' <= char <= 'S'):          
        phone_number += '7'
    elif ('t' <= char <= 'v') or ('T' <= char <= 'V'):          
        phone_number += '8'
    elif ('w' <= char <= 'z') or ('W' <= char <= 'Z'):          
        phone_number += '9'
    else:
        phone_number += '?'

print (f'Phone number is: {phone_number}')
       
```

```python
# Break and continue

# 1. Break statement

for c in "Loki":
    print (c)
print('done!')
print('---')

for c in "Loki":
    if c == 'i':
        break
    print (c)
print('done!')
```

```python
# 2. Continue statement

for c in "Thor":
    if c == 'O':
        continue
    print (c)
print('done!')
print ('---')

for c in "Thor":
    continue
print (c)    
```

```python
# Loop else construct

names = ['Loki', 'Thor', 'Odin', 'heimdall']


num = int(input("Enter a number: "))
print ('Entered number:', num)

for i in range(num):
    if i == num:
        break
    print (names[i], '|', end=' ')
else:
    print("\nLoop completed without break")
```

```python
# loop else 

x = 0
y = 5
z = 5

while x <= y:
    if x == z:
        print('x == z')
    break
    x = x + 1
else:
    print('x == y')
```

```python
# Getting both index and value when looping: enumerate()    
'''Using range() and len() to iterate over a sequence.'''

origins = [2,4,8,10]
for index in range(len(origins)):
    value = origins[index]
    print (f'Element: {index},{value}')
```

```python
# Using list.index() to find the index of each element.

origins = [2,4,8,10]
for value in origins:
    index = origins.index(value)
    print (f'Element: {index},{value}')
```

```python
# The enumerate() function.

origins = [2,4,8,10]
for (index, value) in enumerate(origins):
    print (f'Element: {index},{value}')
```
